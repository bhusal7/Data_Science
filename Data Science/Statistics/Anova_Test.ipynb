```python
import numpy as np 
import seaborn as sns

from scipy.stats import f_oneway
import  pandas as pd 

df = sns.load_dataset('titanic')
df.head()
```

```python
df = df[['age','pclass']].dropna()
df['pclass'].unique()
```

```python
#  creating ckass
class1 = df[df['pclass']==1]['age']
class2 = df[df['pclass']==2]['age']
class3 = df[df['pclass']==3]['age']

print(class1)
```

```python
f_stats , p_value = f_oneway(class1,class2,class3)
f_stats
```

```python
p_value
```

```python
alpha = 0.05
if p_value < alpha:
    print("reject null hypo and ther's significance difference")
else:
    print('There is no difference')
```
