```python
import numpy as np 
import seaborn as sns

from scipy.stats import chi2_contingency
import  pandas as pd 

df = sns.load_dataset('titanic')
df.head()
```

*********** Contingency Table ***************

```python
contingency_table = pd.crosstab(df['sex'],df['survived'])
contingency_table
```

```python
chi2, p_value, dof , expected = chi2_contingency(contingency_table)
print(expected)
print(p_value)
print(chi2)
```

```python
alpha = 0.05

if p_value < alpha:
    print("We reject null hypothesis & there's significant relationship between gender and survivor")
else:
    print("There's no connection ")
```
