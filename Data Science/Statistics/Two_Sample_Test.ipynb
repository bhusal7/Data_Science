```python
import numpy as np
from scipy import stats

groupA = [85,88,90,92,87,85,89,91,86,88]
groupB = [82,84,80,83,81,79,78,85,84,83]

t_stats , p_value = stats.ttest_ind(groupA, groupB, equal_var = False)
print(t_stats)

alpha = 0.25 

if p_value < alpha:
    print("I will reject my null hypothesis")
else:
    print("I will accept my null hypothesis")
```
